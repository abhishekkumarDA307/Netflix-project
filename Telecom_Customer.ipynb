```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv("Telecom_customer_churn.csv")
```

```python
df.head()
```

```python
# As we have 31 Columns Lets sort Columns by their datatype
df.info()
```

```python
df.drop('customerID',axis=1,inplace=True)
```

```python
df.duplicated().sum()  # This will check the duplicate data for all columns.
```

```python
df.drop_duplicates(keep='last',inplace= True)
```

```python
df["TotalCharges"].unique()
```

```python
df[df["TotalCharges"].isin([' ','NA','-'])==True]
```

```python
# Replaceing Whitespaces with null values
df['TotalCharges']= df['TotalCharges'].replace(' ',np.nan)
```

```python
# Converting object datatype into float
df['TotalCharges']= df['TotalCharges'].astype(float)
```

```python
df.isnull().sum()
```

```python
df.dropna(inplace=True)
```

```python
Numerical=df.select_dtypes(exclude="object")
Categorical=df.select_dtypes(include="object")
```

```python
for i in Categorical:
    print(df[i].value_counts())
    print("="*100)
```

```python
sns.set_palette('hsv')
plt.figure(figsize=(30,40), facecolor='white')
plotnumber =1
for i in Categorical:
    if plotnumber <=16:
        ax = plt.subplot(4,5,plotnumber)
        sns.countplot(x=df[i])
        plt.xlabel(i,fontsize=20)
    plotnumber+=1
plt.show()
```

```python
sns.set_palette('husl')
plt.figure(figsize=(12,10))
plt.subplot(1,2,1)
df['Churn'].value_counts().plot.pie(explode=[0,0.1],autopct='%1.2f%%',shadow=True)
plt.title('Churn Distribution', fontsize=22,fontweight ='bold')
plt.subplot(1,2,2)
sns.countplot(x='Churn',data=df)
plt.title('Churn Distribution',fontsize=22,fontweight ='bold')
plt.xlabel("Churn",fontsize=18,fontweight ='bold')
plt.xticks(fontsize=18,fontweight ='bold')
plt.show()
```
